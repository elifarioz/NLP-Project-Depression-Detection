```python
import json
import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split

EATD_DIR = Path("IDAIC/eatd")

with open(EATD_DIR / "mentalHealth_bt.json", encoding="utf-8") as f:
    data = json.load(f)

rows = []
for pid in data["positive_12"]:
    score = float(open(EATD_DIR / "label_files_EATD" / f"{pid}.txt", encoding="utf-8").read().strip())
    text = " ".join(data[f"{q}_12"][pid] for q in ["positive", "neutral", "negative"])
    rows.append({"id": pid, "score": score, "label": int(score >= 53), "text": text.lower()})

df = pd.DataFrame(rows)
print(len(df), "participants |", df["label"].value_counts().to_dict())
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, f1_score, roc_auc_score

df = pd.read_csv("data/eatd.csv")
train = df[df["split"] == "train"]
dev = df[df["split"] == "dev"]
test = df[df["split"] == "test"]

def evaluate(y_true, y_prob, thr=0.5):
    y_pred = (y_prob >= thr).astype(int)
    p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, average="binary", zero_division=0)
    return {
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": p,          # depression class
        "recall": r,             # depression class
        "f1": f,                 # depression class
        "macro_f1": f1_score(y_true, y_pred, average="macro"),
        "auroc": roc_auc_score(y_true, y_prob),
    }
```

```python
vec = TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True)
X_train = vec.fit_transform(train["text"])   # fit on train only
X_dev = vec.transform(dev["text"])
X_test = vec.transform(test["text"])

best = None
for C in [0.1, 1, 10, 100]:
    clf = LogisticRegression(C=C, class_weight="balanced", max_iter=1000, random_state=42)
    clf.fit(X_train, train["label"])
    m = evaluate(dev["label"], clf.predict_proba(X_dev)[:, 1])
    print(f"C={C:<6} dev macro_f1={m['macro_f1']:.3f}  auroc={m['auroc']:.3f}")
    if best is None or m["macro_f1"] > best[1]:
        best = (C, m["macro_f1"], clf)

C, _, clf = best
print("\nBest C:", C)
results = pd.DataFrame({
    "dev": evaluate(dev["label"], clf.predict_proba(X_dev)[:, 1]),
    "test": evaluate(test["label"], clf.predict_proba(X_test)[:, 1]),
}).T.round(3)
results
```

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import RandomizedSearchCV, RepeatedStratifiedKFold, StratifiedKFold, cross_val_predict
from sklearn.metrics import f1_score

train_dev = df[df["split"].isin(["train", "dev"])]
test = df[df["split"] == "test"]

pipe = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("clf", LogisticRegression(solver="liblinear", class_weight="balanced",
                               max_iter=2000, random_state=42)),
])

shared = {
    "tfidf__min_df": [1, 2, 3],
    "tfidf__max_df": [0.8, 0.95, 1.0],
    "tfidf__sublinear_tf": [True, False],
    "tfidf__use_idf": [True, False],
    "clf__penalty": ["l1", "l2"],
    "clf__C": [0.01, 0.1, 1, 10, 100],
}
param_space = [
    {**shared, "tfidf__analyzer": ["word"],
     "tfidf__ngram_range": [(1, 1), (1, 2), (1, 3)],
     "tfidf__stop_words": [None, "english"],
     "tfidf__max_features": [None, 500, 2000]},
    {**shared, "tfidf__analyzer": ["char_wb"],
     "tfidf__ngram_range": [(2, 4), (3, 5)],
     "tfidf__max_features": [None, 5000]},
]

cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=42)
search = RandomizedSearchCV(pipe, param_space, n_iter=120, scoring="roc_auc",
                            cv=cv, random_state=42, n_jobs=-1)
search.fit(train_dev["text"], train_dev["label"])
print("Best CV AUROC:", round(search.best_score_, 3))
print("Best params:", search.best_params_)

# choose the decision threshold from out-of-fold predictions
best = search.best_estimator_
oof = cross_val_predict(best, train_dev["text"], train_dev["label"],
                        cv=StratifiedKFold(5, shuffle=True, random_state=42),
                        method="predict_proba")[:, 1]
thresholds = np.linspace(0.05, 0.95, 91)
scores = [f1_score(train_dev["label"], (oof >= t).astype(int), average="macro") for t in thresholds]
thr = thresholds[int(np.argmax(scores))]
print(f"Best threshold: {thr:.2f} (CV macro_f1={max(scores):.3f})")

# final fit on train + dev, evaluate on test once (uses evaluate() from before)
best.fit(train_dev["text"], train_dev["label"])
pd.DataFrame({"test": evaluate(test["label"], best.predict_proba(test["text"])[:, 1], thr)}).T.round(3)
```

```python
import copy, gc, random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from transformers import AutoTokenizer, AutoModelForSequenceClassification, get_linear_schedule_with_warmup
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, f1_score, roc_auc_score

device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)

def set_seed(seed=42):
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)

# df = pd.read_csv("eatd.csv") 
df = pd.read_csv("data/eatd.csv")
train = df[df["split"] == "train"].reset_index(drop=True)
dev = df[df["split"] == "dev"].reset_index(drop=True)
test = df[df["split"] == "test"].reset_index(drop=True)


def evaluate(y_true, y_prob, thr=0.5):
    y_pred = (np.asarray(y_prob) >= thr).astype(int)
    pc, rc, fc, _ = precision_recall_fscore_support(y_true, y_pred, average=None, labels=[0, 1], zero_division=0)
    wp, wr, wf, _ = precision_recall_fscore_support(y_true, y_pred, average="weighted", zero_division=0)
    return {"accuracy": accuracy_score(y_true, y_pred),
            "nd_precision": pc[0], "nd_recall": rc[0], "nd_f1": fc[0],      # non-depression class
            "precision": pc[1], "recall": rc[1], "f1": fc[1],               # depression class
            "macro_f1": f1_score(y_true, y_pred, average="macro"),
            "weighted_precision": wp, "weighted_recall": wr, "weighted_f1": wf,
            "auroc": roc_auc_score(y_true, y_prob)}


def best_threshold(y_true, y_prob):
    ts = np.linspace(0.05, 0.95, 91)
    scores = [f1_score(y_true, (np.asarray(y_prob) >= t).astype(int), average="macro") for t in ts]
    return ts[int(np.argmax(scores))]
```

```python
def make_chunks(ids, size=510, stride=255):
    """Overlapping windows (50% overlap). 510 tokens + 2 special tokens = 512."""
    if len(ids) <= size:
        return [ids]
    chunks, start = [], 0
    while True:
        chunks.append(ids[start:start + size])
        if start + size >= len(ids):
            break
        start += stride
    return chunks

def build_samples(texts, labels, tokenizer, mode):
    """mode='base': first 510 tokens only. mode='window': all overlapping chunks."""
    cls_id = tokenizer.cls_token_id if tokenizer.cls_token_id is not None else tokenizer.bos_token_id
    sep_id = tokenizer.sep_token_id if tokenizer.sep_token_id is not None else tokenizer.eos_token_id
    samples = []                                    # (input_ids, label, doc_id)
    for doc_id, (text, label) in enumerate(zip(texts, labels)):
        ids = tokenizer(text, add_special_tokens=False)["input_ids"]
        chunks = [ids[:510]] if mode == "base" else make_chunks(ids)
        for ch in chunks:
            samples.append(([cls_id] + ch + [sep_id], int(label), doc_id))
    return samples

def make_collate(tokenizer):
    pad_id = tokenizer.pad_token_id
    def collate(batch):
        max_len = max(len(b[0]) for b in batch)
        input_ids = torch.full((len(batch), max_len), pad_id, dtype=torch.long)
        attention_mask = torch.zeros((len(batch), max_len), dtype=torch.long)
        for i, b in enumerate(batch):
            input_ids[i, :len(b[0])] = torch.tensor(b[0])
            attention_mask[i, :len(b[0])] = 1
        return {"input_ids": input_ids, "attention_mask": attention_mask,
                "labels": torch.tensor([b[1] for b in batch])}
    return collate

def predict(model, samples, n_docs, collate):
    """Probability of depression per document (window: mean over its chunks)."""
    model.eval()
    probs = []
    with torch.no_grad():
        for batch in DataLoader(samples, batch_size=16, shuffle=False, collate_fn=collate):
            inputs = {k: v.to(device) for k, v in batch.items() if k != "labels"}
            probs += torch.softmax(model(**inputs).logits, dim=-1)[:, 1].cpu().tolist()
    doc_ids = np.array([s[2] for s in samples])
    return np.bincount(doc_ids, weights=probs, minlength=n_docs) / np.bincount(doc_ids, minlength=n_docs)

def train_one(model_name, mode, lr, epochs=10, patience=3, batch_size=8, seed=42):
    set_seed(seed)
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    collate = make_collate(tokenizer)
    tr = build_samples(train["text"], train["label"], tokenizer, mode)
    dv = build_samples(dev["text"], dev["label"], tokenizer, mode)
    te = build_samples(test["text"], test["label"], tokenizer, mode)

    model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2).to(device)
    counts = np.bincount(train["label"])
    class_w = torch.tensor(len(train) / (2 * counts), dtype=torch.float).to(device)   # class imbalance
    loss_fn = nn.CrossEntropyLoss(weight=class_w)

    loader = DataLoader(tr, batch_size=batch_size, shuffle=True, collate_fn=collate)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=0.01)
    total = len(loader) * epochs
    sched = get_linear_schedule_with_warmup(opt, int(0.1 * total), total)

    best_auc, best_state, bad = -1, None, 0
    for epoch in range(epochs):
        model.train()
        for batch in loader:
            labels = batch["labels"].to(device)
            inputs = {k: v.to(device) for k, v in batch.items() if k != "labels"}
            loss = loss_fn(model(**inputs).logits, labels)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step(); sched.step(); opt.zero_grad()
        auc = roc_auc_score(dev["label"], predict(model, dv, len(dev), collate))
        if auc > best_auc:
            best_auc, bad = auc, 0
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        else:
            bad += 1
            if bad >= patience:
                break

    model.load_state_dict(best_state)                      # best epoch on dev
    dev_prob = predict(model, dv, len(dev), collate)
    test_prob = predict(model, te, len(test), collate)
    del model, opt; gc.collect(); torch.cuda.empty_cache()
    return best_auc, dev_prob, test_prob
```

```python
MODELS = {"BERT": "bert-base-uncased", "RoBERTa": "roberta-base"}
MODES = ["base", "window"]
LRS = [1e-5, 2e-5, 5e-5]

rows = []
for name, hf_name in MODELS.items():
    for mode in MODES:
        best = None
        for lr in LRS:
            auc, dev_prob, test_prob = train_one(hf_name, mode, lr)
            print(f"{name}-{mode} lr={lr}: dev AUROC={auc:.3f}")
            if best is None or auc > best[0]:
                best = (auc, lr, dev_prob, test_prob)
        auc, lr, dev_prob, test_prob = best
        thr = best_threshold(dev["label"], dev_prob)       # threshold chosen on dev
        row = {"model": f"{name}-{mode}", "lr": lr, "thr": round(thr, 2), "dev_auroc": round(auc, 3)}
        row.update({k: round(v, 3) for k, v in evaluate(test["label"], test_prob, thr).items()})
        rows.append(row)

results = pd.DataFrame(rows)
results.to_csv("eatd_transformer_results.csv", index=False)
results
```

```python
SEEDS = [0, 1, 2, 3, 4]
best_lr = {r["model"]: r["lr"] for _, r in results.iterrows()}   # best lr from the first run

runs = []
for name, hf_name in MODELS.items():
    for mode in MODES:
        key = f"{name}-{mode}"
        for seed in SEEDS:
            auc, dev_prob, test_prob = train_one(hf_name, mode, best_lr[key], seed=seed)
            thr = best_threshold(dev["label"], dev_prob)          # threshold chosen on dev
            m = evaluate(test["label"], test_prob, thr)
            runs.append({"model": key, "seed": seed, **m})
            print(f"{key} seed={seed}: test AUROC={m['auroc']:.3f}  macro_f1={m['macro_f1']:.3f}")

runs_df = pd.DataFrame(runs)
runs_df.to_csv("eatd_multiseed_runs.csv", index=False)

metrics = ["accuracy", "nd_precision", "nd_recall", "nd_f1", "precision", "recall", "f1",
           "macro_f1", "weighted_precision", "weighted_recall", "weighted_f1", "auroc"]
g = runs_df.groupby("model", sort=False)[metrics]
summary = g.mean().round(3).astype(str) + " ± " + g.std().round(3).astype(str)
summary
```
